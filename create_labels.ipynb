```python
import os
import pandas as pd
```

```python
Data_dir = '/data/projects/data/APSS/WP-1/Classification_data/'
all_files = os.listdir(Data_dir)
all_slides = [f for f in all_files if f.endswith('.svs')]
print(f'Total number of slides: {len(all_slides)}')
```

```python
labels = []
for slide_name in all_slides:
    if 'HG' in slide_name or 'Adenocarcinoma' in slide_name or 'Cancer' in slide_name or 'Displasia' in slide_name:
        label = 2
    elif 'Normal' in slide_name or 'Polyp' in slide_name:
        label = 0
    else:
        label = 1
    labels.append({'slide_name': slide_name.split('.')[0], 'label': label})
```

```python
labels_df = pd.DataFrame(labels)
labels_df.to_csv('slide_labels.csv', index=False)
```
